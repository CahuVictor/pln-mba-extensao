Utilizando o spacy, extraia o nome dos personagens presentes no terceiro capitulo da obra "Mémorias postumas de Brás Cubas" de Machado de Assis.

````python
cap_3_bras_cubas = "Mas, já que falei nos meus dois tios, deixem-me fazer aqui um curto esboço genealógico. O fundador de minha família foi um certo Damião Cubas, que floresceu na primeira metade do século XVIII. Era tanoeiro de ofício, natural do Rio de Janeiro, onde teria morrido na penúria e na obscuridade, se somente exercesse a tanoaria. Mas não; fez-se lavrador, plantou, colheu, permutou o seu produto por boas e honradas patacas, até que morreu, deixando grosso cabedal a um filho, o licenciado Luís Cubas. Neste rapaz é que verdadeiramente começa a série de meus avós –– dos avós que a minha família sempre confessou –– porque o Damião Cubas era afinal de contas um tanoeiro, e talvez mau tanoeiro, ao passo que o Luís Cubas estudou em Coimbra, primou no Estado, e foi um dos amigos particulares do vice-rei conde da Cunha. Como este apelido de Cubas lhe cheirasse excessivamente a tanoaria, alegava meu pai, bisneto do Damião, que o dito apelido fora dado a um cavaleiro, herói nas jornadas da Africa, em prêmio da façanha que praticou arrebatando trezentas cubas ao mouros. Meu pai era homem de imaginação; escapou à tanoaria nas asas de um calembour. Era um bom caráter, meu pai, varão digno e leal como poucos. Tinha, é verdade, uns fumos de pacholice; mas quem não é um pouco pachola nesse mundo? Releva notar que ele não recorreu à inventiva senão depois de experimentar a falsificação; primeiramente, entroncou-se na família daquele meu famoso homônimo, o capitão-mor Brás Cubas, que fundou a vila de São Vicente, onde morreu em 1592, e por esse motivo é que me deu o nome de Brás. Opôs-se-lhe, porém, a família do capitão-mor, e foi então que ele imaginou as trezentas cubas mouriscas. Vivem ainda alguns membros de minha família, minha sobrinha Venância, por exemplo, o lírio-do-vale, que é a flor das damas do seu tempo; vive o pai, o Cotrim, um sujeito que... Mas não antecipemos os sucessos; acabemos de uma vez com o nosso emplasto."
cap_3_bras_cubas
````

**<font color='D72009'> Sua resposta aqui </font>**

````python
import spacy

# Carregando modelo de português
nlp = spacy.load("pt_core_news_sm")

texto = cap_3_bras_cubas

doc = nlp(texto)
````

````python
# Forma explícita de fazer

# Lista de POS que você quer **incluir** (deixe vazia para não filtrar)
include_pos = ["PROPN"]

# Lista de POS que você quer **ignorar** (usada se include_pos estiver vazia)
ignore_pos = ["CCONJ", "PUNCT", "SCONJ", "VERB", "ADJ", "NUM"]

# Impressão condicional
for token in doc:
    if include_pos:
        if token.pos_ in include_pos:
            print(f"{token.text:<12} | {token.lemma_:<12} | {token.pos_:<10}")
    else:
        if token.pos_ not in ignore_pos:
            print(f"{token.text:<12} | {token.lemma_:<12} | {token.pos_:<10}")
````

````python
# Criando uma função para reutilização

def filtrar_tokens(doc, include_pos=None, ignore_pos=None):
    """
    Retorna os tokens de um doc com base nas listas de POS fornecidas.

    Parâmetros:
        doc (spacy.tokens.Doc): objeto retornado por nlp(texto)
        include_pos (list[str], opcional): lista de POS a incluir. Se fornecida, só esses serão retornados.
        ignore_pos (list[str], opcional): lista de POS a ignorar. Usada apenas se include_pos for vazia ou None.

    Retorna:
        list[dict]: lista de dicionários com 'text', 'lemma' e 'pos'
    """
    include_pos = include_pos or []
    ignore_pos = ignore_pos or []

    resultado = []

    for token in doc:
        if include_pos:
            if token.pos_ in include_pos:
                resultado.append({
                    "text": token.text,
                    "lemma": token.lemma_,
                    "pos": token.pos_
                })
        else:
            if token.pos_ not in ignore_pos:
                resultado.append({
                    "text": token.text,
                    "lemma": token.lemma_,
                    "pos": token.pos_
                })

    return resultado
````

````python
# Utilizando a função

# Questão 1: incluir apenas substantivos próprios
tokens = filtrar_tokens(doc, include_pos=["PROPN"])

# Impressão formatada
for item in tokens:
    print(f"{item['text']:<12} | {item['lemma']:<12} | {item['pos']:<10}")

"""
# Exemplo 2: ignorar conjunções, pontuação etc.
tokens = filtrar_tokens(doc, ignore_pos=["CCONJ", "PUNCT", "SCONJ", "VERB"])

# Impressão formatada
for item in tokens:
    print(f"{item['text']:<12} | {item['lemma']:<12} | {item['pos']:<10}")
"""
````

````python
# atualizando a função para incorporar o filtro por lemma

def filtrar_tokens(doc, include_pos=None, ignore_pos=None, include_lemmas=None, ignore_lemmas=None):
    """
    Retorna os tokens de um doc com base nas listas de POS e lemas fornecidas.

    Parâmetros:
        doc (spacy.tokens.Doc): objeto retornado por nlp(texto)
        include_pos (list[str], opcional): lista de POS a incluir. Se fornecida, só esses POS serão considerados.
        ignore_pos (list[str], opcional): lista de POS a ignorar. Usada apenas se include_pos estiver vazia ou None.
        include_lemmas (list[str], opcional): lista de lemas a incluir. Se fornecida, apenas esses lemas são considerados.
        ignore_lemmas (list[str], opcional): lista de lemas a ignorar. Usada se include_lemmas estiver vazia ou None.

    Retorna:
        list[dict]: lista de dicionários com 'text', 'lemma' e 'pos'
    """
    include_pos = include_pos or []
    ignore_pos = ignore_pos or []
    include_lemmas = include_lemmas or []
    ignore_lemmas = ignore_lemmas or []

    resultado = []

    for token in doc:
        pos_ok = token.pos_ in include_pos if include_pos else token.pos_ not in ignore_pos
        lemma_ok = token.lemma_ in include_lemmas if include_lemmas else token.lemma_ not in ignore_lemmas

        if pos_ok and lemma_ok:
            resultado.append({
                "text": token.text,
                "lemma": token.lemma_,
                "pos": token.pos_
            })

    return resultado
````

````python
ignore_pos=["ADJ", "ADP", "AUX", "ADV", "CCONJ", "SCONJ", "NUM", "NOUN", "PROPN", "PUNCT", "VERB"]
# ignore_lemmas=["um", "o"]        # ignora essas palavras
ignore_lemmas=[]

# Questão 2: incluir apenas pronomes
tokens = filtrar_tokens(doc, ignore_pos=ignore_pos, ignore_lemmas=ignore_lemmas)

# Impressão formatada
for item in tokens:
    print(f"{item['text']:<12} | {item['lemma']:<12} | {item['pos']:<10}")
````
